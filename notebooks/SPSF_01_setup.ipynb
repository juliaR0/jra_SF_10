```python
### Loading Qwen 2.5 1.5B model
!pip install -q transformers accelerate
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

name = "Qwen/Qwen2.5-1.5B"
tok = AutoTokenizer.from_pretrained(name)
model = AutoModelForCausalLM.from_pretrained(name, torch_dtype=torch.float16).to("cuda")
model.eval()
print("Loaded!", model.config.num_hidden_layers, "layers")
```

```python
### Is model working check
def score(sentence):
    ids = tok(sentence, return_tensors="pt").input_ids.to("cuda")
    with torch.no_grad():
        loss = model(ids, labels=ids).loss
    return -loss.item() * (ids.shape[1] - 1)

for s in ["Yesterday he walked to school.",
          "Yesterday he walk to school.",
          "Ontem ele andou até a escola.",
          "Ontem ele andar até a escola."]:
    print(round(score(s), 2), s)
```
